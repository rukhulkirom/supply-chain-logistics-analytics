## 7. Outlier Analysis

Outlier analysis is performed to identify unusually high or low observations in numerical variables that may require further investigation.

Not all numerical columns are suitable for outlier detection. Identifier columns, binary indicators, categorical codes, and constant variables are excluded because extreme values in these fields do not represent meaningful business outliers.

The analysis focuses on business-related numerical variables such as sales, profit, discounts, quantities, prices, delivery time, and customer-related operational measures.

Statistical outliers are not automatically treated as errors. Extreme observations may represent legitimate business transactions, high-value orders, large quantities, or unusual operational conditions. Any treatment of outliers will therefore be determined during the data cleaning phase based on both statistical evidence and business context.

```python
outlier_columns = [
    "Days for shipping (real)",
    "Days for shipment (scheduled)",
    "Benefit per order",
    "Sales per customer",
    "Order Item Discount",
    "Order Item Discount Rate",
    "Order Item Product Price",
    "Order Item Profit Ratio",
    "Order Item Quantity",
    "Sales",
    "Order Item Total",
    "Order Profit Per Order",
    "Product Price"
]
```

```python
outlier_summary = df[outlier_columns].describe().T

outlier_summary
```

```python
Q1 = df[outlier_columns].quantile(0.25)
Q3 = df[outlier_columns].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
outlier_count = {}

for column in outlier_columns:
    outlier_count[column] = (
        (df[column] < lower_bound[column]) |
        (df[column] > upper_bound[column])
    ).sum()

outlier_summary_iqr = pd.DataFrame({
    "outlier_count": outlier_count
})

outlier_summary_iqr["outlier_percentage"] = (
    outlier_summary_iqr["outlier_count"] / len(df) * 100
).round(2)

outlier_summary_iqr = outlier_summary_iqr.sort_values(
    "outlier_count",
    ascending=False
)

outlier_summary_iqr
```

```python
outlier_analysis = pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower Bound": lower_bound,
    "Upper Bound": upper_bound,
    "Outlier Count": [
        outlier_count[column]
        for column in outlier_columns
    ]
})

outlier_analysis["Outlier Percentage"] = (
    outlier_analysis["Outlier Count"] / len(df) * 100
).round(2)

outlier_analysis
```

## 10. Business Interpretation

### 1. Dataset Structure

The dataset contains 180,519 order-item records and 53 columns covering transactions, customers, products, geography, sales, shipping, and delivery information.

The data is structured at the order-item level rather than the order level. This is supported by the presence of 180,519 unique Order Item Ids compared with 65,752 unique Order Ids.

Therefore, order-level metrics must be calculated using distinct Order Id rather than simply counting rows.

### 2. Delivery Performance

The dataset contains actual shipping duration, scheduled shipping duration, delivery status, and late delivery risk indicators.

The actual shipping duration ranges from 0 to 6 days, while scheduled shipping duration ranges from 0 to 4 days.

The dataset contains 98,977 records with a late delivery risk indicator of 1 and 81,542 records with an indicator of 0.

This indicates that delivery performance is a major analytical area and provides sufficient information to evaluate on-time delivery, delay rate, and shipping efficiency.

### 3. Regional Logistics Analysis

The dataset contains 5 markets, 23 order regions, 164 countries, 1,089 states, and 3,597 cities.

This geographic coverage provides sufficient granularity to compare logistics performance across different regions and identify areas with relatively high delivery delays or longer shipping times.

Regional analysis will therefore be an important component of the supply chain performance evaluation.

### 4. Shipping Mode

The dataset contains four shipping modes: Standard Class, First Class, Second Class, and Same Day.

The availability of both shipping mode and delivery performance variables allows the analysis to compare delivery time and delay rates across different shipping methods.

This can help determine whether certain shipping modes are associated with better or worse delivery performance.

### 5. Product and Category Performance

The dataset contains 50 product categories and 118 unique products.

Product and category information can be linked with delivery performance, allowing the analysis to identify categories or products that consistently experience higher delay rates or longer delivery times.

This provides an opportunity to investigate whether logistics problems are concentrated in specific product groups.

### 6. Sales and Financial Performance

The dataset contains sales, product price, discount, order quantity, and profit-related metrics.

This enables the analysis to connect logistics performance with financial performance.

For example, the project can investigate whether high-value orders experience different delivery performance compared with lower-value orders and whether operational issues may have potential financial implications.

### 7. Customer Impact

The dataset contains customer identifiers, customer segments, geographic information, and order-related information.

Although the dataset does not provide a direct customer satisfaction or customer rating variable, customer segments and order outcomes can still be used to evaluate customer-related patterns.

Customer satisfaction analysis will therefore be limited to variables available in the dataset and will not assume the existence of a direct satisfaction score.

### 8. Data Quality

The dataset contains no exact duplicate rows and no duplicate Order Item Ids.

Most analytical fields are complete. However, several missing-value issues were identified:

- Product Description is completely missing.
- Order Zipcode is missing in approximately 86.24% of records.
- Customer Lname has 8 missing values.
- Customer Zipcode has 3 missing values.

The dataset also passed the main business validation checks, including date consistency, shipping duration validity, order quantity validity, delivery status validity, customer ID consistency, and duplicate Order Item ID checks.

These findings indicate that the dataset is generally suitable for analysis, although some fields should be excluded or handled carefully during data preparation.

### 9. Outlier Interpretation

Several financial variables contain statistical outliers based on the IQR method, particularly Benefit per order, Order Profit Per Order, and Order Item Profit Ratio.

However, these observations are not automatically considered data errors because extreme values may represent legitimate high-value transactions, unusual discounts, or low-margin/high-margin orders.

Therefore, outliers will not be removed automatically. They will be evaluated based on business logic during the data cleaning and preparation phase.

### Overall Business Interpretation

The DataCo Smart Supply Chain Dataset provides sufficient information to perform a comprehensive supply chain and logistics analysis.

The dataset can support analysis of:

- Order volume
- Sales performance
- Delivery time
- Delivery delays
- Shipping mode performance
- Regional logistics performance
- Product and category performance
- Customer patterns
- Profitability

The most important analytical opportunity is to connect operational logistics performance with business outcomes.

The analysis will therefore focus not only on identifying where delays occur, but also on understanding which regions, shipping modes, products, and categories contribute most to delivery problems.

The findings from this phase will guide the data cleaning strategy, feature engineering, PostgreSQL data model, SQL analysis, exploratory data analysis, and Power BI dashboard design in the next phases.
